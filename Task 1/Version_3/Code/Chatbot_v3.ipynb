```python
import pandas as pd
import re
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from transformers import pipeline, AutoModelForCausalLM, AutoTokenizer
import torch
import gradio as gr
from sklearn.preprocessing import normalize

import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score
from rouge_score import rouge_scorer
from bert_score import score as bert_score
```

```python

# Configuration
EMBEDDING_MODEL = "BAAI/bge-small-en-v1.5"
MODEL_NAME = "microsoft/Phi-3.5-mini-instruct"
CHUNK_SIZE = 100
MIN_SIMILARITY = 0.7

# Preprocessing
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)  # Remove special characters
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Load data
data = pd.read_csv("/Parts_processed.csv")
data = data[["ID", "DESCRIPTION"]].dropna()
data['clean_desc'] = data['DESCRIPTION'].apply(preprocess_text)

# Chunking
documents = []
metadata = []
for _, row in data.iterrows():
    text = f"ID {row['ID']}: {row['clean_desc']}"
    for i in range(0, len(text), CHUNK_SIZE):
        chunk = text[i:i+CHUNK_SIZE]
        documents.append(chunk)
        metadata.append({
            "id": row['ID'],
            "full_desc": row['DESCRIPTION'],
            "chunk_num": i // CHUNK_SIZE
        })

# Compute and normalize embeddings
embedder = SentenceTransformer(EMBEDDING_MODEL)
doc_embeddings = embedder.encode(documents, show_progress_bar=True)
doc_embeddings = normalize(doc_embeddings).astype('float32')

# FAISS index (cosine similarity)
index = faiss.IndexFlatIP(doc_embeddings.shape[1])
index.add(doc_embeddings)


id_to_docs = data.set_index('ID')['DESCRIPTION'].to_dict()

# LLM
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    temperature=0.7,
    max_new_tokens=256
)

def find_similar_parts(query_id_or_text, k=5):
    # Check if input is a valid part ID
    if str(query_id_or_text).isdigit() and int(query_id_or_text) in id_to_docs:
        query_text = preprocess_text(id_to_docs[int(query_id_or_text)])
    else:
        query_text = preprocess_text(query_id_or_text)

    # Embed query
    query_embedding = embedder.encode([query_text])
    query_embedding = normalize(query_embedding).astype('float32')

    # FAISS search and results by similarity threshold
    scores, indices = index.search(query_embedding, k * 2)
    results = []
    for score, idx in zip(scores[0], indices[0]):
        if score >= MIN_SIMILARITY:
            results.append({
                "id": metadata[idx]["id"],
                "score": score,
                "description": metadata[idx]["full_desc"]
            })

    # Deduplicate results based on part ID
    seen = set()
    final_results = []
    for r in sorted(results, key=lambda x: -x['score']):
        if r['id'] not in seen:
            seen.add(r['id'])
            final_results.append(r)
            if len(final_results) >= k:
                break

    return final_results[:k]

def generate_answer(query):
    """RAG-based answer for technical parts."""
    similar_parts = find_similar_parts(query)
    if not similar_parts:
        return "No similar parts found based on current data."

    # Build context from similar parts
    context = "\n".join(
        f"Part ID {p['id']} (similarity: {p['score']:.2f}): {p['description']}"
        for p in similar_parts
    )
    prompt = f"""<|system|>
You're a technical parts assistant. List similar parts based on this context:
{context}

<|user|>
Show similar parts for: {query}

<|assistant|>
"""
    response = generator(
        prompt,
        max_new_tokens=512,
        do_sample=True,
        num_beams=3
    )
    return response[0]['generated_text'].split("<|assistant|>")[-1].strip()

def generate_generic_answer(query):
    """Generic chatbot answer using the model's internal knowledge."""
    prompt = f"""<|system|>
You're a knowledgeable and friendly assistant. Provide a clear and concise answer to the user's question using your internal knowledge.

<|user|>
{query}

<|assistant|>
"""
    response = generator(
        prompt,
        max_new_tokens=256,
        do_sample=True,
        num_beams=3
    )
    return response[0]['generated_text'].split("<|assistant|>")[-1].strip()

def combined_generate_answer(mode, query):
    """Selects between RAG and generic chatbot mode."""
    if mode == "Technical Parts Finder":
        return generate_answer(query)
    elif mode == "Chatbot":
        return generate_generic_answer(query)
    else:
        return "Invalid mode selected."


```

```python
# Evaluation
evaluation_data = [
    {
        "query": "give me similar IDs of A1 based on description",
        "reference_answer": "Indicator Red Fast Movement 1.6A 250V Holder Plastic 5 X 20mm Ceramic Box CCC/PSE/VDE/cULus Electric Indicator, Very Fast Blow, 1.6A, 250VAC, 1500A (IR), Inline/holder, 5x20mm"
    },
    {
        "query": "give Red Fast Movement 1.6A 250V Holder Plastic IDs",
        "reference_answer": "Indicator Red Fast Movement 1.6A 250V Holder Plastic 5 X 20mm Ceramic Bulk CCC/CE/CSA/KC/PSE/SEMKO/UL/VDE Electric Indicator, Fast Blow, 1.6A, 250VAC, 1500A (IR), Inline/holder, 5x20mm"
    }
]

def evaluate_generation(evaluation_data):
    bleu_scores = []
    rouge_l_scores = []
    meteor_scores = []
    bert_f1_scores = []
    scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)
    smoothing_function = SmoothingFunction().method1

    for example in evaluation_data:
        query = example["query"]
        reference = example["reference_answer"]
        generated = generate_answer(query)

        # BLEU
        bleu = sentence_bleu([reference.split()], generated.split(), smoothing_function=smoothing_function)
        bleu_scores.append(bleu)

        # ROUGE-L
        rouge_scores = scorer.score(reference, generated)
        rouge_l_scores.append(rouge_scores['rougeL'].fmeasure)


        # BERTScore (F1)
        P, R, F1 = bert_score([generated], [reference], lang="en", verbose=False)
        bert_f1_scores.append(F1[0].item())

    avg_bleu = sum(bleu_scores) / len(bleu_scores)
    avg_rouge_l = sum(rouge_l_scores) / len(rouge_l_scores)
    avg_bert_f1 = sum(bert_f1_scores) / len(bert_f1_scores)

    return avg_bleu, avg_rouge_l, avg_bert_f1

avg_bleu, avg_rouge_l, avg_bert_f1 = evaluate_generation(evaluation_data)
print(f"\ Evaluation Metrics:")
print(f" BLEU: {avg_bleu:.2f}")
print(f" ROUGE-L: {avg_rouge_l:.2f}")
print(f" BERTScore (F1): {avg_bert_f1:.2f}")
```

```python
# gradio interface
def main():
    with gr.Blocks() as demo:
        gr.Markdown("# Technical Parts Finder & Generic Chatbot")
        mode = gr.Radio(choices=["Technical Parts Finder", "Chatbot"], label="Select Mode", value="Technical Parts Finder")
        query = gr.Textbox(label="Enter your query or Part ID/Description")
        output = gr.Textbox(label="Response")
        btn = gr.Button("Submit")
        btn.click(combined_generate_answer, inputs=[mode, query], outputs=output)

    demo.launch()
```

## Project Documentation and Justification  

### Overview  

This project builds a chatbot that helps retrieve technical parts based on their descriptions and also provides general answers. The system is divided into modules: data preprocessing, document chunking, generating embeddings, similarity search, and response generation. We evaluate its performance using BLEU, ROUGE-L, and BERTScore.  

### Model Selection  

- **Primary Model: BAAI/bge-small-en-v1.5**  
  This model gives the best results when finding similar parts and generating responses. It understands language well and captures the details in technical descriptions, making it our first choice.  

- **Alternative Models:**  
  We tested **All-MiniLM-L6-v2** and **all-mpnet-base-v2**, but the BAAI model consistently performed better. The alternatives can still be used for comparison, but they won’t be our main focus.  

- **Why We Didn’t Use LLaMA and Mistral:**  
  These models have gated repositories, which require approvals, and we had limited time to experiment with them.  

- **Future Possibilities:**  
  With a higher budget, we could explore more advanced models like GPT-4, Claude 3 (Anthropic) ,Gemini 1.5  for better response generation.  

### Why We Chose These Libraries  

- **Pandas & NumPy** – For handling data efficiently, especially CSV files and numerical computations.  
- **Regex (`re`)** – For cleaning text by removing special characters and standardizing input.  
- **SentenceTransformers** – For creating embeddings that help find similar descriptions.  
- **FAISS** – For quickly searching large datasets and retrieving relevant parts.  
- **Transformers & PyTorch** – To load and run the **microsoft/Phi-3.5-mini-instruct** model for generating responses.  
- **Gradio** – To create an easy-to-use web interface for testing the chatbot.  
- **Scikit-learn** – To normalize embeddings, making similarity searches more accurate.  
- **NLTK, ROUGE & BERTScore** – To measure how good the chatbot’s answers are.  

### Additional Considerations  

- **General Knowledge Answers:**  
  The chatbot doesn’t just retrieve technical parts—it can also answer general questions. This makes it more useful for broader applications beyond technical searches.  

- **Scalability & Future Enhancements:**  
  The current setup is efficient, but with more computing power and resources, we could integrate larger generative models, improve response accuracy, and expand to multimodal capabilities (e.g., images and text).  

### Conclusion  

By combining these tools and prioritizing the **BAAI/bge-small-en-v1.5** model, i ensure that the chatbot effectively finds the right technical parts and gives useful answers—including general knowledge queries. While we didn’t use LLaMA or Mistral due to time and access constraints, future work could explore more powerful models like GPT-4 for even better performance.
