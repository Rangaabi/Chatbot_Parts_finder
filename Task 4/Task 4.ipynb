Task 4:
Write a function that takes as input two timestamps of the form 2017/05/13 12:00 and 
calculates their differences in hours. Please only return the full hour difference and 
round the results. E.g. 2022/02/15 00:05 and 2022/02/15 01:00 would return 1 hour.

```python
from datetime import datetime, time, timedelta

def calculate_hours_diff(timestamp1, timestamp2):
    """Calculates the rounded hour difference between two timestamps"""
    format_str = "%Y/%m/%d %H:%M"
    dt1 = datetime.strptime(timestamp1, format_str)
    dt2 = datetime.strptime(timestamp2, format_str)
    delta = abs(dt2 - dt1)
    hours = delta.total_seconds() / 3600
    return round(hours)
```

```python
# Testing 1
ts1 = "2022/02/15 00:05"
ts2 = "2022/02/15 01:00"
print("Hour difference:", calculate_hours_diff(ts1, ts2))  

# Testing 2
ts3 = "2025/03/21 15:00"  #Friday
ts4 = "2025/03/24 11:00"  #Monday
print("Hour difference:", calculate_hours_diff(ts3, ts4))  
```
